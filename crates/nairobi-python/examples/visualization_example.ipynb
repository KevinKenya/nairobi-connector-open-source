# Nairobi OS Python Visualization Example

This notebook demonstrates the hardware-accelerated visualization capabilities
of Nairobi OS using Lagos Vision.

````python
# Import the Nairobi OS package
import nairobi_os
import json
import time

print(f"Nairobi OS version: {nairobi_os.__version__}")
print(f"Author: {nairobi_os.__author__}")
````

## Starting the Refinery Daemon

Before we can process data, we need to start the Nairobi Axum Refinery daemon.

````python
# Start the refinery daemon (this will auto-discover the binary)
nairobi_os.start_refinery()

print("Refinery daemon started successfully!")
````

## Available Lagos Visualization Functions

Let's explore what visualization functions are available in the Lagos module.

````python
# Show available visualization functions
lagos_functions = [func for func in dir(nairobi_os.lagos) if not func.startswith('_')]
for func in sorted(lagos_functions):
    print(f"  - {func}")
````
