```python
import nairobi_os
import time

# 1. Ignite the Data Plane
nairobi_os.start_refinery()

print("🚀 Ingesting Heavy Iron Dataset...")
t0 = time.time()
raw_handle = nairobi_os.data.ingest("simulator/PlayerStatisticsExtended.csv")

print("⚙️ Distilling 'points' column via Polars SQL...")
distilled_handle = nairobi_os.data.sql_query(raw_handle, "SELECT points FROM dataset")
print(f"✅ Data distilled in {(time.time() - t0)*1000:.2f} ms")

print("👁️ Spawning Lagos Vision (Jupyter Glass)...")
# 2. Ignite the Visual Cortex
# This spawns the lagos-vision-daemon, connects the WebSocket, and returns the AnyWidget
widget = nairobi_os.lagos.plot_inline(distilled_handle, width=1000, height=400)

# Display the widget
widget
```
